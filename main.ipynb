```python
import pandas as pd
import numpy as np
```

```python
dataset = pd.read_csv('comenzi.csv')
dataset.head()
```

```python
import whisper
model = whisper.load_model('medium')
```

```python
#Modify the name of the audio file to the one you want to transcribe and make sure it's in romanian language
request = model.transcribe('db.mp4', language='ro', fp16=False)
request
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer()
text = vectorizer.fit_transform(dataset['Text'])
request['text'] = vectorizer.transform([request['text']])
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(request['text'], text)
```

```python
import subprocess
import webbrowser
import os

def run_command(command):
    if command == "open_browser":
        webbrowser.open("https://www.google.com")
        print("Actiune: S-a deschis browserul.")
        
    elif command == "open_notepad":
        subprocess.run(["notepad.exe"])
        print("Actiune: S-a deschis Notepad.")
        
    elif command == "exit_assistant":
        print("Actiune: Oprire asistent.")
        exit()
        
    elif command == "search_web":
        webbrowser.open("https://www.google.com")
        print("Actiune: Deschidere browser pentru căutare.")
        
    elif command == "play_music":
        webbrowser.open("https://open.spotify.com")
        print("Actiune: Pornire redare muzică.")
        
    elif command == "get_time":
        import datetime
        ora_curenta = datetime.datetime.now().strftime("%H:%M")
        print(f"Actiune: Ora curentă este {ora_curenta}.")
        
    elif command == "open_youtube":
        webbrowser.open("https://www.youtube.com")
        print("Actiune: S-a deschis YouTube.")
        
    elif command == "open_calculator":
        subprocess.run(["calc.exe"])
        print("Actiune: S-a deschis Calculatorul.")
        
    elif command == "create_folder":
        folder_name = "FolderNou"
        os.makedirs(folder_name, exist_ok=True)
        print(f"Actiune: S-a creat directorul '{folder_name}'.")
        
    elif command == "get_battery":
        print("Actiune: Verificare nivel baterie (indisponibil nativ simplu pe Windows fără pachete externe).")
        
    elif command in ["volume_up", "volume_down", "mute_audio"]:
        print(f"Actiune: Comandă de volum ({command}) primită.")
        
    else:
        print(f"Comanda '{command}' nu este asociata unei actiuni valide.")
```

```python
treshold = 0.3
idx = np.argmax(similarity)
maxx = similarity[idx][0]
if(maxx > treshold):
    command = dataset.iloc[idx]['Command']
    run_command(command)
else:
    print("Comanda nu a fost recunoscuta")
```
